## Minimal test: query a real satellite dataset

Nothing useful here: just ask Earth Engine how many Sentinel-2 images cover Paris over a recent month. If a number comes back, the data connection works end to end.

```python
paris = ee.Geometry.Point([2.3522, 48.8566])

collection = (
    ee.ImageCollection("COPERNICUS/S2_SR_HARMONIZED")
    .filterBounds(paris)
    .filterDate("2024-01-01", "2024-02-01")
)

print(f"Sentinel-2 images found over Paris (January 2024): {collection.size().getInfo()}")
```
