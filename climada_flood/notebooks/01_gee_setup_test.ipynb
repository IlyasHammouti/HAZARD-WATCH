# Test de connexion Google Earth Engine

Objectif : valider que `geemap`/`earthengine-api` sont bien installés et que ton compte GEE fonctionne, avant de commencer la vraie détection satellite.

**Étape manuelle nécessaire :** en exécutant la cellule ci-dessous, une fenêtre navigateur va s'ouvrir pour te demander de te connecter avec ton compte Google et d'autoriser l'accès. Une fois autorisé, tu reviens ici, la cellule se termine toute seule.

**Remplace `TON_PROJECT_ID` ci-dessous par l'ID de ton projet Google Cloud associé à Earth Engine** (visible dans la console Google Cloud, ou dans le code Earth Engine à l'adresse https://code.earthengine.google.com/ en haut à gauche).

```python
import ee

PROJECT_ID = "zeta-bonfire-478712-n9"

ee.Authenticate(auth_mode="localhost")  # ouvre le navigateur, revient automatiquement, sans code à coller
ee.Initialize(project=PROJECT_ID)

print("Connexion Earth Engine OK")
```

## Test minimal : vérifier qu'on peut interroger un vrai jeu de données satellite

On ne fait rien d'utile ici, juste demander à GEE le nombre d'images Sentinel-2 disponibles sur Paris pour une période récente — si ça retourne un nombre, la connexion aux données fonctionne de bout en bout.

```python
paris = ee.Geometry.Point([2.3522, 48.8566])

collection = (
    ee.ImageCollection("COPERNICUS/S2_SR_HARMONIZED")
    .filterBounds(paris)
    .filterDate("2024-01-01", "2024-02-01")
)

print(f"Images Sentinel-2 trouvées sur Paris (janvier 2024) : {collection.size().getInfo()}")
```
